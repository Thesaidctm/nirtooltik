# Quickstart

Uso básico da função `run_experiment` com a configuração de demonstração.

```python
from pathlib import Path
import yaml
from nircls.pipeline import run_experiment

cfg_path = Path("configs/demo.yaml")
with open(cfg_path) as f:
    cfg = yaml.safe_load(f)
cfg["data"]["input_path"] = str((cfg_path.parent / cfg["data"]["input_path"]).resolve())
run_experiment(cfg, Path("results/quickstart"))
```
